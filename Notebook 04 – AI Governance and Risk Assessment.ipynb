## Rules Landscape

This section reviews three major frameworks related to AI governance and risk management: the European Union Artificial Intelligence Act (EU AI Act), the NIST AI Risk Management Framework (AI RMF), and the OECD AI Principles.

The purpose is to understand their main requirements and recommendations and assess their relevance to the ModelGuard project.

These frameworks have different legal and institutional roles. The EU AI Act is a regulatory framework, while NIST AI RMF and the OECD AI Principles provide guidance for responsible AI development and risk management.

The analysis focuses on the current scope of ModelGuard, which is a prototype for customer churn prediction and has not yet been deployed.

### European Union Artificial Intelligence Act (EU AI Act)

The European Union Artificial Intelligence Act (EU AI Act) is a regulatory framework established by the European Parliament and the Council of the European Union. Its main objective is to promote the development and use of trustworthy AI while protecting health, safety, fundamental rights, and other important public interests.

The Act follows a risk-based approach, meaning that AI systems are subject to different requirements depending on their intended purpose and potential risks.

The main regulatory categories include:

- **Prohibited AI Practices:** Certain AI practices that create unacceptable risks are prohibited.
- **High-Risk AI Systems:** Systems classified as high-risk must meet specific requirements related to risk management, data governance, documentation, transparency, human oversight, and other controls.
- **Transparency Requirements:** Certain AI systems are subject to specific transparency obligations.
- **Other AI Systems:** Systems that do not fall into the above categories may still be subject to applicable general requirements.

The classification of an AI system depends on its intended purpose and the conditions specified in the regulation, rather than simply the technology used.


**Applicability to ModelGuard**

ModelGuard uses an XGBoost model to predict customer churn based on customer information. Its intended purpose is to support business decisions by identifying customers who may be at risk of leaving a telecommunications service.

Based on the current project scope, customer churn prediction does not automatically fall into the high-risk categories listed in Annex III of the EU AI Act. However, a definitive legal classification requires examining the actual intended purpose, deployment context, and relevant provisions of the regulation.

The Act may become relevant if the system is placed on the EU market, put into service, or its outputs are used within the European Union under the conditions specified in Article 2.

Since ModelGuard is currently an undeployed academic prototype, this review is a preliminary assessment rather than a formal legal compliance evaluation.

If the system is deployed in a different context or used to support decisions that affect individuals in a more significant way, its regulatory classification should be reviewed again.


### NIST AI Risk Management Framework (AI RMF 1.0)

The National Institute of Standards and Technology (NIST) developed the AI Risk Management Framework (AI RMF 1.0) to help organizations identify, assess, and manage risks associated with artificial intelligence systems.

The framework supports the development of trustworthy AI systems by providing a structured approach to risk management throughout the AI lifecycle.

The NIST AI RMF is organized around four main functions:

1. **GOVERN:** Establishes policies, responsibilities, accountability, and organizational practices for managing AI risks.

2. **MAP:** Identifies the system context, intended purpose, stakeholders, potential impacts, and risks.

3. **MEASURE:** Evaluates and analyzes identified risks using appropriate methods, metrics, and evidence.

4. **MANAGE:** Prioritizes risks and implements appropriate actions to reduce, monitor, or manage them.

The GOVERN function is cross-cutting and supports the other three functions. Risk management is intended to be continuous throughout the AI system lifecycle.



**Applicability to ModelGuard**

The NIST AI RMF is highly relevant to ModelGuard because the project focuses on managing risks and monitoring a machine learning model throughout its lifecycle.

The four NIST functions can be applied as follows:

- **GOVERN:** Define stakeholders, responsibilities, governance policies, and accountability requirements.
- **MAP:** Document the AI system profile, intended purpose, data characteristics, and potential risks.
- **MEASURE:** Evaluate model performance, detect data drift, and assess fairness, explainability, and reliability.
- **MANAGE:** Define risk mitigation actions, monitoring procedures, incident response processes, and model review requirements.

Several activities related to these functions have already been implemented in the previous notebooks. For example, model evaluation was performed in Notebook 01, while experiment tracking and monitoring were addressed in Notebooks 02 and 03.

Notebook 04 will extend this work by defining governance responsibilities, assessing potential risks, and proposing appropriate controls.

The NIST AI RMF will therefore serve as the main operational reference for structuring the ModelGuard risk management process.



### OECD AI Principles

The Organisation for Economic Co-operation and Development (OECD) developed the AI Principles to promote innovative, trustworthy, and human-centered artificial intelligence that respects human rights and democratic values.

The principles were originally adopted in 2019 and updated in May 2024 to address developments in AI technologies and emerging challenges.

The OECD AI Principles include five values-based principles:

1. **Inclusive Growth, Sustainable Development and Well-being:** AI should contribute to economic growth, social well-being, inclusion, and sustainable development.

2. **Respect for the Rule of Law, Human Rights and Democratic Values:** AI systems should respect human rights, fairness, privacy, diversity, and fundamental democratic values.

3. **Transparency and Explainability:** Relevant information about AI systems should be provided to help users and affected individuals understand their outputs and limitations.

4. **Robustness, Security and Safety:** AI systems should remain reliable, secure, and safe throughout their lifecycle, with appropriate mechanisms for managing potential risks.

5. **Accountability:** Organizations and individuals involved in developing, deploying, and operating AI systems should be responsible for their proper functioning and impacts.

In addition to these five values-based principles, the OECD Recommendation includes five recommendations to governments concerning AI research and development, inclusive AI ecosystems, governance environments, human capacity, and international cooperation.



**Applicability to ModelGuard**

The OECD AI Principles provide an ethical and governance foundation for ModelGuard.

Their application to the project can be described as follows:

- **Inclusive Growth and Well-being:** Use churn predictions to support customer retention activities and improve business decision-making.
- **Human Rights and Fairness:** Evaluate whether model performance differs across relevant customer groups and protect customer privacy.
- **Transparency and Explainability:** Document the model's purpose, data sources, limitations, and factors influencing its predictions.
- **Robustness, Security and Safety:** Evaluate model performance and monitor data changes that may affect prediction reliability.
- **Accountability:** Define stakeholder responsibilities and maintain records of model evaluations, risks, incidents, and governance decisions.

The OECD principles will guide the responsible AI considerations within ModelGuard and complement the operational risk management approach provided by NIST AI RMF.

The following table compares the EU AI Act, NIST AI RMF, and OECD AI Principles in terms of their nature, main focus, risk approach, lifecycle coverage, and relevance to ModelGuard.

| Comparison | EU AI Act | NIST AI RMF | OECD AI Principles |

## Impact and Risk Assessment

AI impact and risk assessment helps identify potential negative consequences associated with the development and use of an AI system.

For ModelGuard, the assessment focuses on risks related to data quality, model performance, data drift, fairness, privacy, security, and human decision-making.

Each identified risk will be evaluated based on two factors:

- **Likelihood:** The estimated probability that the risk may occur.
- **Severity:** The potential impact of the risk if it occurs.

Both factors are rated on a scale from 1 to 5. The overall risk score is calculated by multiplying likelihood by severity.

This assessment is preliminary and reflects the current academic prototype and its intended use.

```python
# Risk assessment scale

risk_scale = {
    "1": "Very Low",
    "2": "Low",
    "3": "Moderate",
    "4": "High",
    "5": "Very High"
}

def classify_risk(score):
    if score >= 15:
        return "High"
    elif score >= 6:
        return "Medium"
    return "Low"

print("Risk assessment scale defined successfully.")
print("Likelihood and Severity: 1–5")
print("Risk Score = Likelihood × Severity")
print("Risk classification function is ready.")
```
